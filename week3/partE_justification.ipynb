## 2 · Every parameter I chose, and why  (5 marks)

* **Bin width 2 mmHg (panel 1).** The data are recorded in whole mmHg, so anything below 1 mmHg makes empty bins (in Part A, 400 bins left 296 empty); 3 bins erased the second hump. 2 mmHg is the smallest width that is smooth *and* shows both modes. The split between the two groups is the **valley of a 5-bin moving average (142 mmHg)**.
* **Log y-axis for length of stay (panel 2).** Stay is extremely right-skewed (median 1.4 d, mean 2.3 d, max 273.7 d): on a linear axis every box collapses to a line at the bottom. A log axis is **declared on the panel and in the source note**; **outliers are shown** (`showfliers` left on). Medians are written on the chart. The 2 zero-day stays (both Day Surgery) cannot be drawn on a log axis and are omitted from *this panel only* — the medians are identical with or without them (1.69 / 0.25 / 1.26 / 3.48).
* **Scatter: 2,500-point random subsample, seed 0, alpha 0.25, marker size 9 (panel 3).** 10,000 points are a solid blob; the subsample is declared in the note, and the **line, r and ρ are computed on all 10,000** (not on the subsample). The line is ordinary least squares.
* **Spearman ρ, pairwise-complete, for the heatmap (panel 4).** Spearman is rank-based, so the extreme right tail of stay length and the tied integer values do not dominate (Pearson gives the same picture: 0.34 and 0.89). `cmap="RdBu_r"`, **`vmin = −1, vmax = +1, center = 0`**, `fmt=".2f"`, upper triangle masked. **`map_mmhg` is excluded** because it equals (systolic + 2 × diastolic)/3 (verified to ±0.05), so its correlation with both is arithmetic, not a discovery.
* **Palette:** the course Okabe–Ito subset (blue `#0072B2`, vermilion `#D55E00`) plus a neutral grey; passes Part C (worst CVD pair ΔE ≥ 11). Only two hues are used, so no second channel is needed.
* **Layout and export:** `fig, ax = plt.subplots(2, 2, figsize=(12.6, 8.8))`, spines off, message titles, units on every axis, source note, **`fig.savefig(dpi=300, bbox_inches="tight")` called before `plt.show()`**.
* **Other decisions:** ward spellings conformed by strip → lower → dictionary (12 → 4); slash dates read as **dd/mm/yyyy** (1,933 of 3,253 slash dates have a day > 12, none has a month > 12); `risk_score` is **not** averaged (it is a ratio — e.g. Day Surgery's column mean is 1.42, but its true events ÷ inpatient-days is 0.40).

## 3 · Every row removed: the rule, the count, and what it moved  (4 marks)

| What | Rule | Count | Effect |
|---|---|---|---|
| **Heart-rate device faults** | reading < 30 or > 250 bpm (physiologically impossible at rest; fixed *before* looking at any result) | **n = 37** (29 above 250, up to 415; 8 readings of 0) | mean **78.8 → 78.0 bpm**; SD **20.6 → 13.0** |
| BMI sentinel codes | 999 and 0 are "missing" codes, not measurements | **n = 486** (402 × 999, 84 × 0), set to missing | mean **66.2 → 27.3 kg/m²** (BMI appears only in the heatmap) |
| Zero-day stays | cannot be drawn on a log axis | n = 2, panel 2 only | medians unchanged |
| Ward spellings / dates | conformed, not removed | 6,533 rows relabelled; 0 unparsed | no rows lost |

**The ethics clause.** Readings above 250 bpm were removed as device faults (n = 37 including 8 zero readings). Mean heart rate moves from 78.8 to 78.0 bpm. The figure was also produced without the exclusion; it is in the appendix (`figs/partD_figure_appendix_no_hr_exclusion.png`) and the largest change to any heatmap cell is 0.002.
Heart rate appears in no panel except the heatmap, and removing the faults does not change what it shows (it correlates with nothing either way).

**The symmetry test on my own write-up.** *Would I have made the same removal if it had weakened my conclusion instead of strengthening it?* **Yes.** The 30–250 bpm limits come from physiology (no resting human heart beats at 0 or 415 bpm), were written down before any correlation was computed, apply to every row identically, and are disclosed with the count; with or without them no panel's message changes.
A rule I would have applied only when it helped would be an edit, not a cleaning rule — this one is not.

```python
# Re-derive every number quoted above, so the page can be audited
import numpy as np, pandas as pd
raw = pd.read_csv("data/vitals_10k.csv")
hr_bad = (raw.heart_rate < 30) | (raw.heart_rate > 250)
bmi_bad = raw.bmi.isin([0, 999])
print(f"heart rate: removed {hr_bad.sum()} (>{250}: {(raw.heart_rate > 250).sum()}, zeros: {(raw.heart_rate < 30).sum()}); mean {raw.heart_rate.mean():.1f} -> {raw.heart_rate[~hr_bad].mean():.1f}; SD {raw.heart_rate.std():.1f} -> {raw.heart_rate[~hr_bad].std():.1f}; max {raw.heart_rate.max()}")
print(f"bmi: set to missing {bmi_bad.sum()} (999: {(raw.bmi == 999).sum()}, 0: {(raw.bmi == 0).sum()}); mean {raw.bmi.mean():.2f} -> {raw.bmi[~bmi_bad].mean():.2f}")
print(f"los_days: median {raw.los_days.median():.2f}, mean {raw.los_days.mean():.2f}, max {raw.los_days.max():.2f}; zero-day rows: {(raw.los_days == 0).sum()}")
print(f"systolic < 142: {(raw.systolic < 142).mean():.0%} / >= 142: {(raw.systolic >= 142).mean():.0%}")
```
